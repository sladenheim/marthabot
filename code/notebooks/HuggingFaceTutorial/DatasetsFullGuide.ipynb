#### Process
- HuggingFace Datasets has many tools for modifying structure/content of dataset. Help tidy up dataset, create additiaonl columns, converting b/w features and formats ...
- **For our purposes, clean first!**
- This guide shows you how to:
    - reorder rows, split dataset
    - rename/remove cols, other col operations
    - apply processing functions to each sample in a dataset
    - concatenate datasets
    - apply custom formatting transform
    - save and export processed datasets

- Sort, shuffle, select, split, shard: rearrange structure of datasets. Select only rows you want. Create train/test splits. Shard very large datasets into smaller chunks
- Rename, remove, select, cast, flatten: modify columns. Rename/remove columns. Change columns to new set of features. Flattening nested column structured.
- Map() function. Speed up processing functions --> allows you to apply function to each example in a dataset either independently or in batches. Can create new ros + cols.
- Multiprocessing - parallizes processing on the CPU. Set num_proc parameter. Also works with ranks/indices.
- Batch Processing. Can do with map() function. Size is adjustable. Allows cool stuff like splitting long sentences into shorter chunks + data augmentation.
    - Split long examples
        - Split sentence1 field into chunks of 50 characters
        - Stack all chunks together to create new dataset
- map() for data augmentation.
    - ex.) Generate additional words for a masked token in a sentence.
- Process multiple splits with DatasetDict.map()
- Distributed usage. For map() in a distributed setting, also use torch.distributed.barrier
- batch() - allows for grouping samples from dataset into batches. Useful for batches of data for training/evaluation, espeicall for deep learning models.
- Concatenate - combine separate datasets if same column types shared using concatenate_datasets()
- Interleave - mix several datasets together by alternating examples from each using interleave_datasets(). Custom stopping strategys as well.
- Format - change format of a column using set_format().
- Format transform - apply custom foramtting transform on the fly with set_transform().
    - ex.) tokenize + pad tokens on the fly
- Save and export

### Text Data

#### Load text data
- Text files are one of most common file types for storing a dataset.
    - By default, HuggingFace Datasets samples txt file line by line to build dataset.
    - ex.) dataset = load_dataset("text", data_files={"train": ["my_text_1.txt", "my_text_2.txt"], "test": "my_test_file.txt"})
    - ex.) dataset = load_dataset("text", data_dir="path/to/text/dataset")
- Can sample text file by paragraph or by entire document, using sample_by parameter
    - Sample by paragraph:
        - ex.) dataset = load_dataset("text", data_files={"train": "my_train_file.txt", "test": "my_test_file.txt"}, sample_by="paragraph")
    - Sample by document:
        - ex.) dataset = load_dataset("text", data_files={"train": "my_train_file.txt", "test": "my_test_file.txt"}, sample_by="document")
- Grep Patterns to load specific files
- Load remote text files via HTTP --> pass URLs instead
- Load XML data using xml loader.

#### Process text data
- Tokenize dataset with map()
- Align dataset labels with label ids for NLI datasets
    - *NLI = Natural language inference = task of determining whether the given hypothesis logically follows from the premise. TO understand relationship b/w two pieces of text. 
- **Map**
    - map() function supports processing batches of examples which speeds up tokenization.
    - batched = True
    - map() converts returned values to a PyArrow supported-format but returning tensors as NumPy arrays is faster. Set return_tensors = 'np'
- **Align**
    - align_labels_with_mapping() aligns dataset label id with label name.
