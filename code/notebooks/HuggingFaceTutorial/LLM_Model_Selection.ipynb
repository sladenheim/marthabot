# **General Task Notes**
### **We are looking to fine-tune a base model**
- **What is a base model of an LLM?**
  - A **large, pre-trained neural network**.

- **Fine-tuning Definition:**
  - Process of taking a **pre-trained LLM** and **continuing its training** on a smaller, new dataset to perform a **specialized task**.
  - Updates **some or all** of the model’s weights.

### **Our Current Dataset Format**
- **Each row = one sentence of writing**.
- **First column = prompt (first half of sentence)**.
- **Second column = answer (second half of sentence)**.
- **Goal:** **Predict the second half based on the first half**.

---

# **What Type of NLP Task is This?**
### **Language Modeling (Causal/Autoregressive)**
- Need a model to **predict the next word(s) in a sequence**.
- Generates text **token-by-token, left to right** → **autoregressive**.

### **Conditional Text Generation**
- **Generates text output based on input** → suggests a **Seq2Seq task**.

---

# **What Type of NLP Task is This on Hugging Face?**
- **Hugging Face categorizes models based on tasks.**
- **General Needs for MarthaBot:**
  - **CLM** → **GPT-2/3, LLaMA 2, Mistral 7B** (often used for text generation).
  - **Text Generation** → **GPT-4, Falcon, Mixtral**.

### **Related Hugging Face Tasks**
#### **Text Generation (aka Causal Language Modeling)**
- **Predicts the next word** given the previous words/tokens/input text.
- **Predicts one token at a time** → results in **natural, human-like text**.
- Typically **decoder-only / autoregressive models**.
- **Better for chatbots!**
- **Example Models:** **GPT, BLOOM, PaLM**.

#### **Text2Text Generation**
- **Transforms one type of text into another**.
- Model is trained to **map input text to output text**.
- **Processes the entire input, then generates output**.
- Best for **structured sentence/text transformation tasks**.
- Typically **Seq2Seq architecture (encoder-decoder)**.
- **Example Models:** **T5, BART, NLLB, FLAN-T5**.

# **Common Families of LLMs**
## **Different Architecture Types of LLMs**
### **Autoregressive Models**
- **Generate text sequentially**, predicting **one token at a time**, conditioned on previous tokens.
- Uses **probability distribution** to select the most likely **next word/character**.
- **Strengths:**  
  - **Excels at fluent and contextually appropriate text generation**.
  - **Best for open-ended text generation tasks**.
- **Weaknesses:**  
  - Left-to-right generation **may struggle with long-term coherence**.
- **Example Models:** **GPT (Generative Pre-Trained Transformer)**.

---

### **Autoencoding Models**
- **Grasp/understand context of words in a sentence** by predicting **masked tokens**.
- Uses **bidirectional context** (analyzes both left & right context when making predictions).
- **Strengths:**  
  - **Deep understanding of context & semantics**.
  - Ideal for **sentiment analysis, question answering, NER**.
- **Weaknesses:**  
  - **Cannot generate text freely**.
  - **Expensive bidirectional attention → slower inference**.
- **Example Models:** **BERT, RoBERTa, ALBERT, DistilBERT, ELECTRA**.

---

### **Seq2Seq Models**
- Designed for tasks where **input & output are sequences of varying lengths**.
- **Encoder-decoder approach:**
  - **Encoder** → compresses input into **latent vector**.
  - **Decoder** → generates **target sequence one token at a time**.
- **Strengths:**  
  - **Machine translation, text summarization, text generation, question-answering**.
- **Weaknesses:**  
  - **Expensive**.
  - **Struggles with long-range dependencies**.
  - **Exposure bias**.
- **Example Models:** **T5, BART**.

---

All model families have **strengths, weaknesses, and various applications**.

## **LLaMA Family (Meta AI)**
- **Highly efficient, open-source models**.
- **Autoregressive, decoder-only Transformer architecture**.
- **Processes tokens sequentially**.
- **Uses causal self-attention** → tokens only attend to past tokens.
- **Optimizations:**
  - **GQA (Grouped Query Attention)** → Speeds up inference.
  - **RoPE (Rotary Positional Embeddings)** → Captures long-range dependencies.
  - **Pre-normalization** → Improves training stability.
- **LLaMA 2:**
  - **Comparable to GPT-3.5 but more lightweight & easier to deploy**.
  - **Best open-source GPT alternative**.
- **Use Cases:**
  - **Academic research**.
  - **AI Chatbots**.
  - **Deployable on local machines**.

---

- **Gold standard for high-performing open-source models**.
- **Decoder/Autoregressive architecture**.
- **Mixtral = Mixture of Experts model**.
- **Outperforms LLaMA 2 with fewer parameters**.
- **Strong text generation, Q&A, summarization**.
