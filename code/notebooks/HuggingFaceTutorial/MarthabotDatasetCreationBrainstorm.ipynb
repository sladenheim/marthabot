**How to handle formal writing vs Question/Answer data**
- Approach 1: Two Separate Datasets
    - Dataset 1: Prose Dataset --> formal writing from Martha (books/essays/articles).
    - Dataset 2: Interview/Q and A Dataset --> transcripts from Q&A or conversational interviews.
    - Challenge:
        - Do we then train two different models? How do we combine them later?
        - Could we use one dataset as a base for writing then the interview dataset to get conversational tone?
        - Could we manually choose if chatbot answers in narrative vs conversational output?
- Approach 2: One Dataset + Style Tag/Way to distinguish
    - Tag each text block with prose or interview
    - ex.) {"text": "<prose> Something about dancing </prose>"},
    - At generation time:
        - Use the tag as a prompt.
        - Some models do well with tags.
- Approach 3: Q & A as Dialogue. Conditional Style Prompting
    - More advacned idea
    - Make prose the default writing style
    - Use Q&A as a condtional geenration task
    - Resembles dialogue modeling
    - OFten uses encoder-decoder model or fine-tuning LLaMA style models with conversational formats
